```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, TimeDistributed
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split
from collections import Counter
import re
import pickle
import matplotlib.pyplot as plt
from tqdm import tqdm
import os

EMB_SIZE = 100
CONTEXT_LEN = 10
LSTM_UNITS = 128
BATCH_SIZE = 128
EPOCHS = 10
TEST_SPLIT = 0.2
RANDOM_STATE = 42

TEXT_FILE = "Harry_Potter_cleaned.txt"
W2V_MODEL_DIR = "word2vec_saved"
LSTM_MODEL_FILE = "lstm_lm_model.h5"
VOCAB_FILE = "vocab.pkl"

os.makedirs(W2V_MODEL_DIR, exist_ok=True)

class Word2Vec:
    def __init__(self, vocab_size, emb_size):
        self.vocab_size = vocab_size
        self.emb_size = emb_size
        self.W_in = np.random.uniform(-0.5/emb_size, 0.5/emb_size, (vocab_size, emb_size)).astype(np.float32)
        self.W_out = np.random.uniform(-0.5/emb_size, 0.5/emb_size, (vocab_size, emb_size)).astype(np.float32)

    def train_batch(self, targets, pos_list, negs_list, lr=0.01):
        batch_targets = []
        batch_contexts = []
        batch_labels = []
        for t, pos, negs_per_pos in zip(targets, pos_list, negs_list):
            for p, negs in zip(pos, negs_per_pos):
                batch_targets.append(t)
                batch_contexts.append(p)
                batch_labels.append(1)
                for n in negs:
                    batch_targets.append(t)
                    batch_contexts.append(n)
                    batch_labels.append(0)

        target_idx = np.array(batch_targets, dtype=np.int64)
        context_idx = np.array(batch_contexts, dtype=np.int64)
        labels = np.array(batch_labels, dtype=np.float32)

        target_vecs = self.W_in[target_idx]
        context_vecs = self.W_out[context_idx]
        scores = np.sum(target_vecs * context_vecs, axis=1)
        sig = 1 / (1 + np.exp(-np.clip(scores, -10, 10)))
        loss = - (labels * np.log(sig + 1e-10) + (1 - labels) * np.log(1 - sig + 1e-10)).mean()

        d_scores = sig - labels
        d_context = d_scores[:, np.newaxis] * target_vecs
        d_target = np.zeros_like(self.W_in)
        np.add.at(d_target, target_idx, d_scores[:, np.newaxis] * context_vecs)

        np.add.at(self.W_out, context_idx, -lr * d_context)
        self.W_in -= lr * d_target
        return loss

def train_word2vec(text_file, emb_size, window=3, neg_samples=10, epochs=7, batch_size=512):
    print("Обучение word2vec...")
    with open(text_file, "r", encoding="utf-8") as f:
        words = f.read().split()
    print("Всего слов:", len(words))

    min_count = 5
    word_counts = Counter(words)
    vocab = [w for w, c in word_counts.items() if c >= min_count]
    word2idx = {w: i for i, w in enumerate(vocab)}
    idx2word = {i: w for w, i in word2idx.items()}
    freqs = np.array([word_counts[w] for w in vocab], dtype=np.float32)
    vocab_size = len(vocab)
    print("Размер словаря:", vocab_size)

    probs = freqs ** 0.75
    probs /= probs.sum()

    idx_seq = [word2idx[w] for w in words if w in word2idx]

    def generate_batches(idx_seq, window, neg_samples, probs, batch_size):
        batch_targets = []
        batch_pos = []
        for i, target in enumerate(idx_seq):
            left = max(0, i - window)
            right = min(len(idx_seq), i + window + 1)
            pos = idx_seq[left:i] + idx_seq[i+1:right]
            if not pos:
                continue
            batch_targets.append(target)
            batch_pos.append(pos)
            if len(batch_targets) >= batch_size:
                negs_list = []
                for t, pos_list in zip(batch_targets, batch_pos):
                    pos_negs = []
                    for _ in pos_list:
                        negs = np.random.choice(vocab_size, size=neg_samples, replace=True, p=probs).tolist()
                        pos_negs.append(negs)
                    negs_list.append(pos_negs)
                yield batch_targets, batch_pos, negs_list
                batch_targets = []
                batch_pos = []
        if batch_targets:
            negs_list = []
            for t, pos_list in zip(batch_targets, batch_pos):
                pos_negs = []
                for _ in pos_list:
                    negs = np.random.choice(vocab_size, size=neg_samples, replace=True, p=probs).tolist()
                    pos_negs.append(negs)
                negs_list.append(pos_negs)
            yield batch_targets, batch_pos, negs_list

    model = Word2Vec(vocab_size, emb_size)
    for epoch in range(epochs):
        total_loss = 0.0
        num_batches = 0
        pbar = tqdm(total=len(idx_seq), desc=f"Эпоха {epoch+1}", unit="слов")
        for targets, pos_list, negs_list in generate_batches(idx_seq, window, neg_samples, probs, batch_size):
            loss = model.train_batch(targets, pos_list, negs_list)
            total_loss += loss
            num_batches += 1
            pbar.update(len(targets))
        pbar.close()
        avg_loss = total_loss / num_batches if num_batches else 0
        print(f"Средняя потеря: {avg_loss:.4f}")

    return model, word2idx, idx2word, vocab_size

if os.path.exists(os.path.join(W2V_MODEL_DIR, "W_in.npy")) and os.path.exists(VOCAB_FILE):
    print("Загрузка сохранённой модели word2vec...")
    with open(VOCAB_FILE, 'rb') as f:
        vocab_data = pickle.load(f)
        word2idx = vocab_data['word2idx']
        idx2word = vocab_data['idx2word']
        vocab_size = vocab_data['vocab_size']
    w2v_model = Word2Vec(vocab_size, EMB_SIZE)  # временный объект
    w2v_model.W_in = np.load(os.path.join(W2V_MODEL_DIR, "W_in.npy"))
    w2v_model.W_out = np.load(os.path.join(W2V_MODEL_DIR, "W_out.npy"))
    w2v_model.vocab_size = w2v_model.W_in.shape[0]
    w2v_model.emb_size = w2v_model.W_in.shape[1]
else:
    w2v_model, word2idx, idx2word, vocab_size = train_word2vec(TEXT_FILE, EMB_SIZE, epochs=10)
    # Сохраняем веса и словари
    np.save(os.path.join(W2V_MODEL_DIR, "W_in.npy"), w2v_model.W_in)
    np.save(os.path.join(W2V_MODEL_DIR, "W_out.npy"), w2v_model.W_out)
    with open(VOCAB_FILE, 'wb') as f:
        pickle.dump({'word2idx': word2idx, 'idx2word': idx2word, 'vocab_size': vocab_size}, f)
    print("Модель word2vec сохранена.")

with open(TEXT_FILE, "r", encoding="utf-8") as f:
    text = f.read()

sentences = re.split(r'[.,!?;:-]', text)
sentences = [s.strip() for s in sentences if len(s.strip()) > 0]
print(f"Количество кусочных предложений: {len(sentences)}")

word_sequences = []
for sent in sentences:
    words = sent.split()
    seq = [word2idx.get(w, 0) for w in words if w in word2idx]
    if len(seq) >= CONTEXT_LEN + 1:
        word_sequences.append(seq)

print(f"Предложений достаточной длины: {len(word_sequences)}")

X_data, y_data = [], []
for seq in tqdm(word_sequences, desc="Создание выборок"):
    for i in range(len(seq) - CONTEXT_LEN):
        X_data.append(seq[i:i+CONTEXT_LEN])
        y_data.append(seq[i+1:i+CONTEXT_LEN+1])

X_data = np.array(X_data, dtype=np.int32)
y_data = np.array(y_data, dtype=np.int32)
print(f"Размер выборки: {X_data.shape}")

X_train, X_test, y_train, y_test = train_test_split(
    X_data, y_data, test_size=TEST_SPLIT, random_state=RANDOM_STATE
)
print(f"Обучающая выборка: {X_train.shape}, тестовая: {X_test.shape}")
embedding_matrix = w2v_model.W_in

# ОНО ВОТ ТУТ САМОЕ ИНТЕРЕСНОЕ

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=EMB_SIZE, weights=[embedding_matrix],
              trainable=True, mask_zero=False),
    LSTM(LSTM_UNITS, return_sequences=True),
    LSTM(LSTM_UNITS, return_sequences=True),
    TimeDistributed(Dense(vocab_size, activation='softmax'))
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()

history = None
if os.path.isfile(LSTM_MODEL_FILE):
    print("Загружаю модель...")
    model = tf.keras.models.load_model(LSTM_MODEL_FILE)
else:
    history = model.fit(
        X_train, y_train,
        batch_size=BATCH_SIZE,
        epochs=EPOCHS,
        validation_data=(X_test, y_test),
        verbose=1
    )

    model.save(LSTM_MODEL_FILE)
    print(f"Модель LSTM сохранена в {LSTM_MODEL_FILE}")

test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
perplexity = np.exp(test_loss) # test_loss - кросс энтропия !!!
print(f"\nТестовая accuracy: {test_acc:.4f} ({test_acc*100:.2f}%)")
print(f"Тестовый perplexity: {perplexity:.2f}")

if history is not None:
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    plt.plot(history.history['loss'], label='Train loss')
    plt.plot(history.history['val_loss'], label='Val loss')
    plt.title('Loss')
    plt.xlabel('Epoch')
    plt.legend()
    plt.subplot(1,2,2)
    plt.plot(history.history['accuracy'], label='Train acc')
    plt.plot(history.history['val_accuracy'], label='Val acc')
    plt.title('Accuracy')
    plt.xlabel('Epoch')
    plt.legend()
    plt.tight_layout()
    plt.savefig('lstm_training_curves.png')
    plt.show()

def predict_next_word(model, context_words, word2idx, idx2word, context_len=CONTEXT_LEN):
    context_indices = [word2idx.get(w, 0) for w in context_words]
    if len(context_indices) < context_len:
        context_indices = [0] * (context_len - len(context_indices)) + context_indices
    else:
        context_indices = context_indices[-context_len:]
    X = np.array([context_indices], dtype=np.int32)
    preds = model.predict(X, verbose=0)
    last_step_probs = preds[0, -1, :]
    predicted_idx = np.argmax(last_step_probs)
    return idx2word.get(predicted_idx, "<UNK>")

print("\nПримеры предсказаний:")
test_examples = [
    ["гарри", "поттер", "и", "его"],
    ["гермиона", "взяла", "книгу"],
    ["волан", "де", "морт", "вернулся"],
    ["хогвартс", "это", "школа"],
    ["снейп", "преподавал", "зельеварение"],
    ["драко", "малфой", "был"]
]
for example in test_examples:
    next_word = predict_next_word(model, example, word2idx, idx2word)
    print(f"Контекст: {' '.join(example)} -> следующее слово: {next_word}")

from google.colab import drive
import os

# Монтируем Google Drive
drive.mount('/content/drive')

# Путь для сохранения на Диске
DRIVE_SAVE_PATH = '/content/drive/MyDrive/Colab_Models/HarryPotter'
os.makedirs(DRIVE_SAVE_PATH, exist_ok=True)
print(f"Директория на Диске готова: {DRIVE_SAVE_PATH}")
import shutil

def copy_to_drive(file_or_dir, target_dir):
    if os.path.exists(file_or_dir):
        if os.path.isdir(file_or_dir):
            dest = os.path.join(target_dir, os.path.basename(file_or_dir))
            if os.path.exists(dest): shutil.rmtree(dest)
            shutil.copytree(file_or_dir, dest)
            print(f"Папка {file_or_dir} скопирована на Диск.")
        else:
            shutil.copy(file_or_dir, target_dir)
            print(f"Файл {file_or_dir} скопирован на Диск.")
    else:
        print(f"Ошибка: {file_or_dir} не найден.")

# Копируем все ключевые файлы
copy_to_drive('word2vec_saved', DRIVE_SAVE_PATH)
copy_to_drive('vocab.pkl', DRIVE_SAVE_PATH)
copy_to_drive('lstm_lm_model.h5', DRIVE_SAVE_PATH)
copy_to_drive('lstm_training_curves.png', DRIVE_SAVE_PATH)
```
